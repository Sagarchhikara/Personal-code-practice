## Data Visualization 

```python
# import the required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
```

```python
# Curve ploting 
x = np.linspace(0, 5, 50)
y_cos = np.cos(x)
y_sin = np.sin(x)

plt.plot(x, y_cos, 'r--')

plt.plot(x, y_sin)

plt.xlabel('X')
plt.ylabel('Y')
plt.title('Sin and Cos Curves')
plt.show()

```

```python
#Using Pannels
x = np.linspace(0, 5, 50)
y_cos = np.cos(x)
y_sin = np.sin(x)


plt.subplot(2, 2, 1)
plt.plot(x, y_cos, 'r--')

plt.title('cos')
plt.subplot(2, 2, 4)
plt.plot(x, y_sin, 'b-')
plt.title('sin')
plt.show()


# Some references for further personalizations:
# http://matplotlib.org/api/lines_api.html#matplotlib.lines.Line2D.set_linestyle
# http://matplotlib.org/api/colors_api.html
# http://matplotlib.org/api/markers_api.html
# http://matplotlib.org/api/lines_api.html#matplotlib.lines.Line2D
# It is possible to have customer colors by RGB values (in range [0-1])
# plt.plot(x,y_sin,'b-',color = (0.1,0.9,0.9))
# The curve's color will be cyan
```

```python
# scatter plots
from sklearn.datasets import make_blobs

Dataset = make_blobs(n_samples = 100, n_features = 2,
                    centers =  3, random_state = 7)

Y = Dataset[1]
X = Dataset[0]

plt.plot(X[Y==0, 0], X[Y==0, 1],\
         'ys', label = 'Group 0' )

plt.plot(X[Y==1, 0], X[Y==1, 1],\
         'm*', label = 'Group 1' )
plt.plot(X[Y==2, 0], X[Y==2, 1],\
         'rD', label = 'Group 2' )
plt.legend(loc = 'lower left', numpoints = 1)

plt.ylim(-2,10)
plt.yticks([10, 6, 2, -2])
plt.xticks([-15, -5, 5, 15])

plt.annotate('Squares', (-10, 2))
plt.annotate('Stars', (0, 2))
plt.annotate('Diamonds', (8, 4))
plt.show()
```

```python
#histograms
x = np.random.normal(loc = 0, scale = 1, size =500 )
z = np.random.normal(loc = 8, scale = 1, size =500 )
plt.hist(np.column_stack((x,z)), bins = 20, 
        histtype = 'bar', color = ['c', 'b'], 
         stacked = False)
```

```python
# BAr graphs
X = np.arange(5)
y1 = [34, 56, 12, 68, 67]
y2 = [30, 50, 10, 60, 60]
y3 = [40, 60, 15, 70, 75]
width = 0.2
plt.bar(X-0.2, y1, width, color = 'cyan')
plt.bar(X, y2, width, color = 'orange')
plt.bar(X+0.2, y3, width, color = 'green')
```

```python
# Bar graphs 
from sklearn.datasets import load_iris

iris = load_iris()
average = np.mean(iris.data, axis = 0)
std = np.std(iris.data, axis = 0)
range_ = range(np.shape(iris.data)[1])

plt.subplot(1,2,1)
plt.title('Horizontal Bars')
plt.barh(range_, average, color = 'r', xerr = std, 
         align = 'center', alpha = 0.2)
plt.yticks(range_, iris.feature_names)

plt.subplot(1,2,2)
plt.title('Vertical Bars')
plt.bar(range_, average, color = 'b', yerr = std, 
         align = 'center', alpha = 0.2)
plt.xticks(  range_, iris.feature_names, rotation = 45)
```

```python
#Image Visualization 
from sklearn.datasets import fetch_olivetti_faces

dataset = fetch_olivetti_faces(shuffle = True,
                               random_state = 5)

for k in range(6):
    plt.subplot(2,3, k+1)
    plt.imshow(dataset.data[k].reshape(64, 64), 
              cmap = plt.cm.gray, 
              interpolation = 'nearest')
    plt.title('Face '+str(dataset.target[k]))
    plt.axis('off')
plt.show()
```

```python
#image Visualization: Hand written digits
from sklearn.datasets import load_digits
digits = load_digits()
for number in range(1, 10):
    fig = plt.subplot(3,3, number)
    plt.imshow(digits.images[number],cmap = 'binary', 
              interpolation = 'none', extent  = [0, 8, 0, 8])
    fig.set_xticks(np.arange(0, 9, 1))
    fig.set_yticks(np.arange(0, 9, 1))
    fig.grid()

plt.show()
```

```python
# Selected Graphical Examples with Pandas

from sklearn.datasets import load_iris
iris = load_iris()
iris_df = pd.DataFrame(iris.data, 
                       columns = iris.feature_names)
groups = list(iris.target)
iris_df['groups'] = pd.Series([iris.target_names[k] for k in groups])
```

```python
boxplot = iris_df.boxplot(return_type = 'axes')
```

```python
densityplot = iris_df.plot(kind = 'density')
```

```python
Singlw_dist = iris_df['petal width (cm)'].plot(kind = 'hist', alpha = 0.4)
```

```python
# Scatter plot using dataframes
colors_plt = {0:'red', 1:'yellow', 2:'blue'}
colors = [colors_plt[c] for c in groups]
iris_df.plot(kind = 'scatter', x = 2, y = 3, c = colors)
```

```python
# hexbin using data frames
iris_df.plot(kind = 'hexbin', x = 0, y= 1, gridsize = 10)
```

```python
# Scatter Matrix
from pandas.plotting import scatter_matrix
colors_plt = {0:'red', 1:'green', 2:'blue'}
colors = [colors_plt[c] for c in groups]
scatter_matrix(iris_df, alpha = 0.7, 
              figsize=(6, 6), color = colors, 
              diagonal = 'kde')
```

```python
# Seaborn
import seaborn as sns
sns.set()
# Curve ploting 
x = np.linspace(0, 5, 50)
y_cos = np.cos(x)
y_sin = np.sin(x)
plt.plot(x, y_cos)
plt.plot(x, y_sin)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Sin and Cos Curves')
plt.show()

```

```python
iris = load_iris()
X  = iris.data
Y = iris.target
features = [a[:-5].replace(' ','_') 
            for a in iris.feature_names]
targets = {j: flower for j, flower in enumerate(iris.target_names)}
iris_df = pd.DataFrame(X, columns = features)
iris_df['target'] = [targets[y] for y in Y]

with sns.axes_style('whitegrid'):
    chart = sns.pairplot(data = iris_df, 
                         hue = "target",
                         diag_kind = 'hist')
```

```python
## Visulization
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
import  matplotlib.pyplot as plt
X1, Y1 = make_regression(n_samples=100, n_features = 1,\
                         noise = 5)
model = LinearRegression()
model.fit(X1, Y1)
Y1_pred = model.predict(X1)

plt.scatter(X1, Y1)
plt.plot(X1, Y1_pred, 'r-')
```
